#### I can change the item in a list, i can use the methods .append and .insert to add an item to the end of the list, or insert a new value in the given loaction and move everything to the right. .insert requires 2 parameters, the index you want to insert it at and what the value is.

```python
activities = ["bowling", "cooking", "flying", "relaxing", "walking", "gaming"]
print(activities)

activities.remove("cooking")
print(activities)
activities.pop(4)
print(activities)
```

#### you can remove items from a list with .remove and .pop.   .remove requires you to use the exact value while .pop uses the index

```python
numbers = [3, 5, 8, 10, 11, 28, 42, 77, 101, 108, 2002]
numbers[2] = 1
numbers.insert(2, 50)
numbers.remove(108)
numbers.append(100)
numbers.pop(-2)
numbers.pop(5)
del(numbers[1])
print(numbers)

```
